```python
import re

def normalize_arabic(text):

    # أ، إ، آ، ا → ا
    text = re.sub(r"[إأآا]", "ا", text)

    # ى → ي
    text = re.sub(r"ى", "ي", text)

    # ة → ه
    text = re.sub(r"ة", "ه", text)

    # ؤ → و
    text = re.sub(r"ؤ", "و", text)

    # إزالة التطويل ـ
    text = re.sub(r"ـ+", "", text)

    return text


text = "يأإآا لبيه، على مـؤسسة كبيرة، وفيها مدرسة جميلة!"

print("Before:", text)
print("After :", normalize_arabic(text))
```

```python
texts = [
    "يا زين هالجو، ودي أطلع للبر اليوم!",
    "وش رأيك نطلب عشا ونشوف لنا فلم؟",
    "ترى الزحمة اليوم مو طبيعية أبد"
]


for text in texts:
    print("Before:", text)
    print("After :", normalize_arabic(text))
    print("-" * 40)
```
